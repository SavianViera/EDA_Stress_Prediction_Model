# EDA Pipeline — NeuroKit2 Highpass Decomposition

Drop-in replacement for the cvxEDA notebook using NeuroKit2's highpass filter method.

| Original (cvxEDA) | This notebook (highpass) |
|---|---|
| `CVX_Phasic` | `HP_Phasic` |
| `CVX_Tonic` | `HP_Tonic` |
| `CVX_Tonic.csv` | `HP_Tonic.csv` |
| `Tonic_Reshaped.csv` | `HP_Tonic_Reshaped.csv` |
| `Final_Result.csv` | `HP_Final_Result.csv` |

```python
import pandas as pd
import numpy as np
!pip install neurokit2
import neurokit2 as nk
```

```python
df = pd.read_csv("MatLab_EDA_For_R.csv")
df.head()
```

## 1. Decompose into Phasic and Tonic (NeuroKit2 highpass)

```python
processed_r_columns = []

for col_name in df.columns:
    current_column_series = df[col_name].values.flatten()

    signals, info = nk.eda_process(current_column_series, sampling_rate=4)
    r = signals['EDA_Phasic'].values

    r_series = pd.Series(r, index=df.index, name=f'Phasic_high_pass_{col_name}')
    processed_r_columns.append(r_series)

HP_Phasic = pd.concat(processed_r_columns, axis=1)
HP_Phasic.head()
```

```python
processed_t_columns = []

for col_name in df.columns:
    current_column_series = df[col_name].values.flatten()

    signals, info = nk.eda_process(current_column_series, sampling_rate=4)
    t = signals['EDA_Tonic'].values

    t_series = pd.Series(t, index=df.index, name=f'tonic_high_pass_{col_name}')
    processed_t_columns.append(t_series)

HP_Tonic = pd.concat(processed_t_columns, axis=1)
HP_Tonic.head()
```

```python
HP_Phasic.to_csv('HP_Phasic.csv', index=False)
HP_Tonic.to_csv('HP_Tonic.csv',   index=False)
```

## 2. Assign condition labels and reshape Tonic into 10-second windows

```python
def assign_condition_by_row(row_index):  # 1-based, matching original R code
    if 1    <= row_index <= 239:  return "neutral"
    if 240  <= row_index <= 539:  return "black"
    if 540  <= row_index <= 779:  return "positive"
    if 780  <= row_index <= 1075: return "black"
    if 1076 <= row_index <= 1300: return "negative"
    return None


def reshape_by_condition(df_input, window_size=10, sampling_rate=4):
    chunk_size = window_size * sampling_rate
    data_cols  = [c for c in df_input.columns if c != 'condition']
    cond_col   = df_input['condition'].reset_index(drop=True)
    data       = df_input[data_cols].reset_index(drop=True)

    block_starts = [0]
    for i in range(1, len(cond_col)):
        if cond_col[i] != cond_col[i - 1]:
            block_starts.append(i)
    block_starts.append(len(cond_col))

    rows_out, index_labels = [], []

    for b in range(len(block_starts) - 1):
        start = block_starts[b]
        end   = block_starts[b + 1]
        cond  = cond_col[start]
        block = data.iloc[start:end]
        n_chunks = int(np.ceil(len(block) / chunk_size))

        for c in range(n_chunks):
            cs    = c * chunk_size
            ce    = min(cs + chunk_size, len(block))
            chunk = block.iloc[cs:ce]
            src   = list(range(start + cs + 1, start + ce + 1))

            for col in data_cols:
                vals   = chunk[col].values
                padded = np.full(chunk_size, np.nan)
                padded[:len(vals)] = vals

                row = {f'V{i+1}': padded[i] for i in range(chunk_size)}
                row['condition']   = cond
                row['source_rows'] = ', '.join(str(r) for r in src)
                rows_out.append(row)
                index_labels.append(f'{col}_{(c+1)*window_size}s')

    result = pd.DataFrame(rows_out, index=index_labels)
    result['condition'] = pd.Categorical(
        result['condition'],
        categories=["neutral", "positive", "negative", "black"]
    )
    return result


HP_Tonic['condition'] = [assign_condition_by_row(i + 1) for i in range(len(HP_Tonic))]
HP_Tonic = HP_Tonic[HP_Tonic['condition'].notna()].copy()
HP_Tonic['condition'] = pd.Categorical(
    HP_Tonic['condition'],
    categories=["neutral", "positive", "negative", "black"]
)

hp_result = reshape_by_condition(HP_Tonic)
hp_result.head()
```

```python
hp_result.to_csv('HP_Phasic_Reshaped.csv', index=True)
```

## 3. Summary statistics per window

```python
numeric_result = hp_result.drop(columns=['condition', 'source_rows'])

hp_phasic_desc = pd.DataFrame({
    'Mean_Phasic': numeric_result.mean(axis=1),
    'Max_Phasic':  numeric_result.max(axis=1),
    'Min_Phasic':  numeric_result.min(axis=1),
    'SD_Phasic':   numeric_result.std(axis=1)
})

print(hp_phasic_desc)
```

```python
numeric_result = hp_result.drop(columns=['condition', 'source_rows'])

hp_tonic_desc = pd.DataFrame({
    'Mean_Tonic': numeric_result.mean(axis=1),
    'Max_Tonic':  numeric_result.max(axis=1),
    'Min_Tonic':  numeric_result.min(axis=1),
    'SD_Tonic':   numeric_result.std(axis=1)
})

print(hp_tonic_desc)
```

## 4. Merge with raw features

```python
Final_Dataframe = pd.read_csv("Raw_Features.csv")

hp_tonic_desc_for_merge = hp_tonic_desc.copy()
hp_tonic_desc_for_merge['X'] = hp_tonic_desc_for_merge.index.str.replace('tonic_high_pass_', '', regex=False)
hp_result_merged = pd.merge(Final_Dataframe, hp_tonic_desc_for_merge, on='X', how='left')
```

```python
hp_phasic_desc_for_merge = hp_phasic_desc.copy()
hp_phasic_desc_for_merge['X'] = hp_phasic_desc_for_merge.index.str.replace('Phasic_high_pass_', '', regex=False)
hp_result_merged = pd.merge(hp_result_merged, hp_phasic_desc_for_merge, on='X', how='left')

hp_result_merged.head()
```

```python
hp_result_merged.to_csv('HP_Final_Result.csv', index=False)
```
