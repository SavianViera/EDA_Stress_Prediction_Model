```python
# Initialize an empty list to store the 'r' results for each column
processed_r_columns = []

# Iterate over each column name in the DataFrame df
for col_name in df.columns:
  # Select the current column by its name
  current_column_series = df[col_name]

  # Apply cvxEDA to the current column Series
  # Assuming 'r' is the primary output of interest to collect for each column
  r, _, _, _, _, _, _ = cvxeda.cvxEDA(current_column_series, 1.0 / 4)

  # Create a pandas Series from the 'r' output, using the original index for alignment
  # and a meaningful column name (e.g., 'r_SUD_003_01_EDA')
  r_series = pd.Series(r, index=df.index, name=f'Phasic_{col_name}')

  # Add this processed Series to our list
  processed_r_columns.append(r_series)

# After processing all columns, concatenate the list of Series into a new DataFrame
# This will create a DataFrame where each column is the 'r' component for an original df column
CVX_Phasic = pd.concat(processed_r_columns, axis=1)

# You might want to display the head of the new DataFrame to check the result
CVX_Phasic.head()
```

```python
# Initialize an empty list to store the 'r' results for each column
processed_t_columns = []

# Iterate over each column name in the DataFrame df
for col_name in df.columns:
  # Select the current column by its name
  current_column_series = df[col_name]

  # Apply cvxEDA to the current column Series
  # Assuming 'r' is the primary output of interest to collect for each column
  _, _, t, _, _, _, _ = cvxeda.cvxEDA(current_column_series, 1.0 / 4)

  # Create a pandas Series from the 'r' output, using the original index for alignment
  # and a meaningful column name (e.g., 'r_SUD_003_01_EDA')
  t_series = pd.Series(t, index=df.index, name=f'tonic_{col_name}')

  # Add this processed Series to our list
  processed_t_columns.append(t_series)

# After processing all columns, concatenate the list of Series into a new DataFrame
# This will create a DataFrame where each column is the 'r' component for an original df column
CVX_Tonic = pd.concat(processed_t_columns, axis=1)

CVX_Tonic.head()
```

```python
import itertools
import math
import pandas as pd

# ---------------------------------------------------------------------------
# Condition assignment using markers (matches Function_For_Window_And_Overlap...)
# ---------------------------------------------------------------------------
markers = [
    60,           # end of n
    2 * 60 + 15,  # end of b
    3 * 60 + 15,  # end of p
    4 * 60 + 29,  # end of b
    5 * 60 + 26,  # end of negative
]
markers = [m * 4 for m in markers]


def assign_condition_by_row(row_index):
    """row_index is 1-based, matching R's indexing."""
    if 1 + 240 <= row_index <= markers[0] - 1 + 240:
        return "neutral"
    if markers[0] + 240 <= row_index <= markers[1] - 1 + 240:
        return "black"
    if markers[1] + 240 <= row_index <= markers[2] - 1 + 240:
        return "positive"
    if markers[2] + 240 <= row_index <= markers[3] - 1 + 240:
        return "black"
    if row_index >= markers[3] + 240:
        return "negative"
    return None


condition_levels = ["neutral", "positive", "negative", "black"]


def label_conditions(df_input):
    df_input = df_input.copy()
    df_input["condition"] = [
        assign_condition_by_row(i) for i in range(1, len(df_input) + 1)
    ]
    df_input = df_input[df_input["condition"].notna()].reset_index(drop=True)
    df_input["condition"] = pd.Categorical(
        df_input["condition"], categories=condition_levels
    )
    return df_input


CVX_Phasic = label_conditions(CVX_Phasic)
CVX_Tonic = label_conditions(CVX_Tonic)


# ---------------------------------------------------------------------------
# Reshape function (unchanged)
# ---------------------------------------------------------------------------
def reshape_by_condition(df_input, window_size=10, overlap=0):
    chunk_size = window_size * 4
    colnames_data = [c for c in df_input.columns if c != "condition"]
    conditions = df_input["condition"].astype(str).tolist()

    values, lengths = [], []
    for val, grp in itertools.groupby(conditions):
        values.append(val)
        lengths.append(len(list(grp)))

    block_ends_count = list(itertools.accumulate(lengths))
    block_starts = [0] + block_ends_count[:-1]

    chunk_list = []
    for start, end_count, cond in zip(block_starts, block_ends_count, values):
        end = end_count - 1
        row_indices = list(range(start, end + 1))
        n_rows = len(row_indices)
        n_chunks = math.ceil(n_rows / chunk_size)
        padded = row_indices + [None] * (n_chunks * chunk_size - n_rows)
        chunks = [padded[j * chunk_size:(j + 1) * chunk_size] for j in range(n_chunks)]
        chunk_list.append({"chunks": chunks, "condition": cond, "n_chunks": n_chunks})

    if overlap > 0:
        for i in range(len(chunk_list) - 1):
            current_block = chunk_list[i]
            next_block_start = block_starts[i + 1]
            next_block_end = block_ends_count[i + 1] - 1

            overlap_end = min(next_block_start + overlap - 1, next_block_end)
            overlap_indices = (
                list(range(next_block_start, overlap_end + 1))
                if overlap_end >= next_block_start else []
            )

            last_chunk = current_block["chunks"][-1]
            real_indices = [x for x in last_chunk if x is not None]
            combined = real_indices + overlap_indices

            if len(combined) <= chunk_size:
                last_chunk_new = combined + [None] * (chunk_size - len(combined))
            else:
                last_chunk_new = combined

            chunk_list[i]["chunks"][-1] = last_chunk_new

    all_chunks, all_conditions = [], []
    for block in chunk_list:
        all_chunks.extend(block["chunks"])
        all_conditions.extend([block["condition"]] * block["n_chunks"])

    n_cols2 = len(all_chunks)

    if n_cols2 == 0:
        print("Warning: n_cols2 is 0, no column renaming will occur.")

    rows, row_labels = [], []
    for col_name in colnames_data:
        vals = df_input[col_name].to_numpy()
        for chunk_idx, idx_list in enumerate(all_chunks):
            out = [None] * max(chunk_size, len(idx_list))
            for pos, idx in enumerate(idx_list):
                if idx is not None:
                    out[pos] = vals[idx]
            out = out[:chunk_size]
            rows.append(out)
            row_labels.append(f"{col_name}_{(chunk_idx + 1) * window_size}s")

    value_colnames = [f"t{j + 1}" for j in range(chunk_size)]
    df_reshaped = pd.DataFrame(rows, index=row_labels, columns=value_colnames)

    df_reshaped["condition"] = pd.Categorical(
        all_conditions * len(colnames_data), categories=condition_levels
    )

    source_rows_str = [
        ", ".join(str(idx + 1) for idx in idx_list if idx is not None)
        for idx_list in all_chunks
    ]
    df_reshaped["source_rows"] = source_rows_str * len(colnames_data)

    return df_reshaped


# ---------------------------------------------------------------------------
# Build reshaped Phasic and Tonic dataframes
# ---------------------------------------------------------------------------
result_phasic = reshape_by_condition(CVX_Phasic)
result_tonic = reshape_by_condition(CVX_Tonic)
```
